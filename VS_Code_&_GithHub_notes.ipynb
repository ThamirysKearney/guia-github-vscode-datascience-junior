Esta guía está diseñada para que trabaje con las mejores prácticas de la industria, desde la organización de mis carpetas hasta la colaboración en equipo.

### 1.2 Crear archivos esenciales: ¿Qué es lo que más usaré? 📊

En el día a día de un Data Analyst, usarás archivos específicos para que tu trabajo sea escalable y profesional.

**Atajos para crear archivos:**
- **Desde el ratón**: Icono de "New File" en la barra lateral.
- **Desde teclado**: `Ctrl + N` crea un archivo sin nombre. Luego dale a `Ctrl + S` para ponerle nombre y extensión (ej. `mi_archivo.py`).

| Archivo profesional | Comando `touch` | Uso común del Data Analyst |
| :--- | :--- | :--- |
| **`analysis.py`** | `touch analysis.py` | **El Motor**: Aquí va la lógica pesada, cálculos y KPIs. |
| **`config.py`** | `touch config.py` | **El Cerebro**: Centraliza rutas de carpetas, IDs de provincias y parámetros. |
| **`app.py`** (o `pipeline.py`) | `touch app.py` | **El Conductor**: Llama a los demás archivos en el orden correcto. |
| **`requirements.txt`** | `touch requirements.txt` | **La Lista**: Todas las librerías necesarias (`pandas`, `matplotlib`). |
| **`.gitignore`** | `touch .gitignore` | **El Escudo**: Evita subir basura (`__pycache__`) o archivos personales. |
| **`.env`** | `touch .env` | **Caja Fuerte**: Aquí guardamos las llaves API y contraseñas (NUNCA se sube a GitHub). |
| **`EDA.ipynb`** | `touch EDA.ipynb` | **El Laboratorio**: Para exploración inicial y gráficos rápidos. |

## 🛠️ PARTE 2: Flujo Maestro de Git
### 2.1 Iniciar el repositorio local 🏠
```bash
git init
```
**⚠️ Importante**: Este comando solo ocurre en **MI ordenador**. Crea una carpeta oculta `.git`. No tiene conexión con GitHub hasta que se la indiques. Es tu "máquina del tiempo" privada.

### 2.2 Guardar mis cambios (El ciclo de oro) 🔄
Si trabajas en equipo (como con Mariana y Rokio), el ciclo cambia ligeramente para no pisar el trabajo de los demás.

```bash
# 1. ANTES DE NADA: Trae lo que el equipo ha hecho
git pull origin develop

# 2. Trabaja en tus archivos...

# 3. Prepara los cambios
git add . 

# 4. Haz el commit (la foto de tus cambios)
git commit -m "feat: add human-readable labels to analysis engine"
```
**💡 Buenas Prácticas**: Este flujo es universal. Da igual si trabajas en **GitHub**, **GitLab** o **Bitbucket**; los comandos son los mismos. Es el lenguaje secreto de todos los desarrolladores del mundo.

### 2.3 Conectar con GitHub (Subir mis cambios a la nube) ☁️

| Concepto | ¿Cuándo se usa? | Comando |
| :--- | :--- | :--- |
| **`git clone`** | Cuando el proyecto **YA EXISTE** en GitHub y quieres una copia. | `git clone [URL]` |
| **`git remote add`** | Cuando tienes el código en tu PC y quieres **CREAR** el repo en GitHub. | `git remote add origin [URL]` |

**¿Por qué me pides `push -u origin main`?**
El `-u` (upstream) crea un vínculo permanente. Solo se hace la primera vez. 

**⚠️ Regla de Oro Senior**: Nunca subirás nada directamente a `main` en una empresa. Usarás ese comando para tu rama: 
`git push -u origin feature/mi-mejora`

## 🌿 PARTE 3: Ramas y Colaboración Profesional
Las ramas son "universos paralelos". Nos permiten trabajar sin romper la versión que funciona.

1. **Crear y entrar en rama**: `git checkout -b feature/nombre-tarea` 
2. **Para subir tus cambios a develop**:
   - Haz el `push` de tu rama: `git push origin feature/nombre-tarea`.
   - Ve a la web de GitHub y abre un **Pull Request**.

### 🔄 Diferencia Vital: Merge vs Pull Request
- **Merge a MI RAMA**: Traigo los cambios de `develop` a mi trabajo para estar al día. (Acción técnica).
- **Merge a DEVELOP**: Integro mi trabajo terminado en el proyecto común. (Acción de equipo en GitHub).

**Diferencia de flujo**:
- **En local**: `git merge develop` (Tú traes lo de ellas).
- **En GitHub**: Mariana le da al botón "Merge Pull Request" (Ellas aceptan lo tuyo).

## ⚠️ PARTE 4: ¡Socorro! Emergencias de Git (Los trucos del chat)

| Lo que quieres hacer | El comando mágico 🧙‍♂️ |
| :--- | :--- |
| **Quitar un archivo de Git pero NO borrarlo del PC** | `git rm --cached nombre_archivo` |
| **Corregir el mensaje del último commit** | `git commit --amend -m "nuevo mensaje"` |
| **Añadir un archivo que olvidaste al último commit** | `git add [archivo]` seguido de `git commit --amend --no-edit` |
| **Sobreescribir historial en GitHub (con cuidado)** | `git push origin [rama] --force` |
| **Deshacer `git add .`** | `git reset` |

## 📋 PARTE 5: Referencia Rápida de Comandos Bash
Los analistas pasan mucho tiempo en la terminal. ¡Domínala!

| Comando Bash | Acción | Tip de Analista |
| :--- | :--- | :--- |
| `pwd` | Ver ruta actual | Úsalo si `pandas` no encuentra un CSV. |
| `ls -lh` | Ver archivos y su tamaño | Ideal para ver cuánto pesan tus CSV/Parquet. |
| `cat archivo` | Ver contenido rápido | Mira las primeras líneas de un archivo sin abrir VS Code. |
| `mkdir -p A/B/C` | Crear carpetas anidadas | Crea `data/raw` y `data/processed` de un solo golpe. |
| `echo "..." > f` | Crear archivo con contenido | Crea un `.gitignore` rápido desde la terminal. |

### Resumen de Git para la Daily
- **`git fork`**: Copias el proyecto de otra persona SU cuenta a TU cuenta de GitHub (para practicar).
- **`git merge`**: Une dos ramas. Se usa al final de una tarea.
- **`git status`**: Tu brújula. Úsala cada 5 minutos.
