# 🚀 VS Code & GitHub: Guía Definitiva para Data Analysts

Esta guía unifica la mejor información en **un solo archivo completo, actualizado y estructurado**, con las mejores prácticas de Git y VS Code. Funcionará como tu manual de referencia maestro.

## 📂 PARTE 1: Preparación del Terreno (VS Code + Terminal)

### 1.1 Abrir VS Code y la terminal
- Abre VS Code.
- Si no ves la terminal abajo, presiona: ``Ctrl + ` `` o ve a **Terminal -> New Terminal**.
- Asegúrate de que la terminal esté en **Bash** (o PowerShell si trabajas en Windows).

> ⚠️ En Git Bash (WSL/MinGW) las rutas son estilo UNIX: `/c/Users/USUARIO/...`.
> En PowerShell/Windows usa: `C:\Users\USUARIO\...`

### 1.2 Crear una carpeta desde cero (estructura profesional)
```bash
# 1. Mira en qué carpeta estás ahora
pwd

# 2. Ve al Escritorio (o la carpeta donde quieras trabajar)
cd ~/Desktop

# 3. Crea la carpeta del proyecto
mkdir proyecto-data-analyst

# 4. Entra en esa carpeta
cd proyecto-data-analyst

# 5. Abre VS Code en esa carpeta
code . -r
```

### 1.3 Archivos esenciales y comandos rápidos
Para que tu trabajo sea escalable y profesional, así se crean los archivos rápidamente con `touch`:

| Archivo | Propósito | Comando en Terminal |
| :--- | :--- | :--- |
| `README.md` | Documentación del proyecto | `touch README.md` |
| `requirements.txt` | Librerías necesarias | `touch requirements.txt` |
| `.gitignore` | Evita subir basura (pycache, .env) | `touch .gitignore` |
| `.env` | Variables secretas (nunca subir) | `touch .env` |
| `config.py` | Parámetros y rutas centralizadas | `touch config.py` |
| `analysis.py` | Lógica principal y cálculos | `touch analysis.py` |
| `notebook.ipynb` | Exploración de datos y experimentos| `touch notebook.ipynb` |
| `datos.csv` | Almacenamiento local de datos | `touch datos.csv` |

### 1.4 Cambiar el nombre de un archivo
Si necesitas renombrar un archivo, usa el comando `mv` (move):
```bash
mv borrador.ipynb analisis_final.ipynb
```
*💡 Consejo: Esto funciona para cualquier tipo de archivo (.py, .js, .csv, etc).*

## 🛠️ PARTE 2: Flujo Maestro de Git (Paso a Paso)

### 2.1 Iniciar el repositorio (Solo 1 vez por proyecto)
```bash
git init
```

### 2.2 Guardar tus cambios (El ciclo que usarás siempre)
```bash
# 1. Ver qué ha cambiado
git status

# 2. Preparar el archivo (ponerlo en la "caja de salida")
git add .                    # Añade todo lo nuevo/modificado

# 3. Crear el Commit (Guardado permanente con nombre claro)
git commit -m "feat: initial data cleaning of sales dataset"
```

### 2.3 Conectar con GitHub (Subir tus cambios a la nube)
Crea un repo vacío en GitHub y conecta tu PC:
```bash
# 1. Conecta tu PC con la nube (una sola vez)
git remote add origin https://github.com/tu-usuario/tu-repo.git

# 2. Sube tu rama principal por primera vez
git push -u origin main
```
> ⚠️ **Recuerda**: En equipos se trabaja con ramas; no se hace `push` directo a `main`.

## 🌿 PARTE 3: Ramas y Colaboración Profesional

Nunca trabajes directamente en `main`. Crea ramas para experimentos o nuevas fases.

### 3.1 Crear tu rama de trabajo
```bash
git checkout -b feature/mi-tarea
```

### 3.2 Trabajar y subir tu avance (Formato Profesional)
Para subir tus cambios a `develop` o `main` de forma profesional, sigue este orden:

```bash
# PRIMERO: Traer lo último (evita conflictos)
git pull origin develop

# SEGUNDO: Haz tus commits en tu rama
git add .
git commit -m "feat: descripción de lo que hice"

# TERCERO: Sube tu rama
git push origin feature/mi-tarea
```

### 3.3 Abrir y Procesar un Pull Request
1. Ve a tu repositorio en GitHub.
2. Crea un **Pull Request** desde tu rama (`feature/mi-tarea`) hacia `develop`.
3. Solicita revisión. Cuando te den el OK, haz el **Merge**.

## ⚠️ PARTE 4: Emergencias de Git (¡Socorro!)

Aquí tienes cómo solucionar los errores más comunes. Recuerda: en Git casi todo tiene marcha atrás.

| Quiero... | Comando | Qué hace |
| :--- | :--- | :--- |
| **Editar último mensaje** | `git commit --amend -m "nuevo mensaje"` | Cambia solo el mensaje del último commit. |
| **Deshacer `git add`** | `git reset nombre_archivo` | Lo saca de staging, sin borrar tus cambios locales. |
| **Deshacer un PUSH** | `git revert HEAD` | Crea un commit nuevo que "anula" el último cambio en GitHub. Es lo más seguro en equipo. |
| **Deshacer un PULL** | `git reset --hard ORIG_HEAD` | Si todo se rompe tras un pull, vuelves a como estabas. |
| **Deshacer un MERGE** | `git merge --abort` | Cancela el merge en curso por un conflicto y deja todo igual. |
| **Añadir archivo olvidado** | `git add archivo && git commit --amend --no-edit` | Mete el archivo al último commit sin cambiar el mensaje. |

### Explicación rápida de "Resets"
- **`Soft Reset` (`git reset --soft HEAD~1`)**: Quita el commit pero tus archivos no cambian. Muy útil para re-empaquetar un commit.
- **`Hard Reset`**: Borra el historial Y los cambios de tus archivos. ¡Peligroso! Úsalo sólo si estás seguro.
- **`Revert`**: La opción más madura.

### ¿Qué pasa con un submódulo?
Para eliminar ese submódulo y mantener todo en el repo padre:
```bash
# 1. Quitar el puntero del submódulo (no borra archivos locales)
git rm --cached -r nombre-del-submodulo

# 2. Borrar la carpeta .git del submódulo
rm -rf nombre-del-submodulo/.git

# 3. Añadir la carpeta como archivos normales
git add nombre-del-submodulo

# 4. Commit y push
git commit -m "chore: remove submodule and track as normal files"
```

## 📝 PARTE 6: Nomenclatura Profesional de Commits

Como Analista de Datos, siempre usamos en la industria este formato: `tipo: descripción corta en inglés`.

| Tipo | Cuándo usarlo | Ejemplo |
| :--- | :--- | :--- |
| `feat:` | Nueva funcionalidad / gráfico | `feat: build sales bar chart` |
| `fix:` | Corregir un error / métrica | `fix: correct null handling in parser` |
| `docs:` | Cambios en documentación | `docs: update setup steps in README` |
| `data:` | Modificar datasets o carga | `data: load Q3 financial records` |
| `chore:` | Tareas rutinarias (limpieza) | `chore: clean repo structure` |
| `refactor:`| Optimizar código ya hecho | `refactor: simplify data loader loop` |
